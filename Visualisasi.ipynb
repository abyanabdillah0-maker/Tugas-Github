```python
import matplotlib.pyplot as plt
import seaborn as sns

# Data
bulan = ['Jan', 'Feb', 'Mar', 'Apr', 'Mei', 'Juni', 'Juli']
penjualan = [150, 200, 180, 250, 125, 200, 130]

plt.figure(figsize=(8, 4)) # Diperlebar sedikit agar 7 bulan muat dengan rapi

# 1. Buat Grafik Batang
sns.barplot(x=bulan, y=penjualan, palette='muted')

# 2. Tambahkan Garis Tren (Menghubungkan puncak batang)
# range(len(bulan)) menghasilkan [0, 1, 2, 3, 4, 5, 6] sebagai posisi sumbu X
plt.plot(
    range(len(bulan)),  # Posisi X (tengah setiap batang)
    penjualan,          # Posisi Y (nilai penjualan)
    color='red',        # Warna garis
    marker='o',         # Tambahkan titik bulat di setiap persimpangan
    linestyle='-',      # Gaya garis (solid)
    linewidth=2,        # Ketebalan garis
    label='Garis Tren'  # Nama untuk legenda
)

# 3. Kustomisasi Tampilan
plt.title('Penjualan per Bulan dengan Garis Tren', fontsize=12, fontweight='bold')
plt.xlabel('Bulan')
plt.ylabel('Jumlah Penjualan')
plt.legend() # Menampilkan kotak legenda

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Data
bulan = ['Jan', 'Feb', 'Mar', 'Apr', 'Mei', 'Juni', 'Juli']
penjualan = [150, 200, 180, 250, 125, 220, 130]

# 2. Mengubah ke DataFrame dan Mengurutkan (Terkecil ke Terbesar)
df = pd.DataFrame({'bulan': bulan, 'penjualan': penjualan})
df_sorted = df.sort_values(by='penjualan', ascending=True) # ascending=True

# 3. Membuat Grafik
plt.figure(figsize=(8, 5)) # Sedikit dilebarkan agar label bulan tidak tumpang tindih

# Plot menggunakan data yang sudah diurutkan (ditambahkan hue & legend=False untuk menghindari warning)
ax = sns.barplot(data=df_sorted, x='bulan', y='penjualan', hue='bulan', palette='muted', legend=False)

plt.title('Penjualan per Bulan', fontsize=12, fontweight='bold')
plt.xlabel('Bulan')
plt.ylabel('Jumlah Penjualan')

# 4. Menambahkan Nilai di Ujung Grafik (Modern & Rapi)
# ax.containers[0] mengambil semua objek batang (bar) dari grafik
for container in ax.containers:
    ax.bar_label(
        container,
        padding=3, # Jarak teks dari ujung batang
        fontsize=9,
        fontweight='bold',
        color='#333333' # Warna font abu-abu gelap
    )

# 5. Menampilkan
plt.tight_layout()
plt.show()
```

#Visualisasi dataset 1

```python
from matplotlib import pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

dataa1 = pd.read_csv('https://raw.githubusercontent.com/Alam-A99/Pandas_Latih/10ebb3af000e930a44989075733f495d5c2f4be5/penjualan.csv')
dataa1
```

```python
# 2. Tambahkan kolom total untuk memudahkan beberapa grafik
dataa1['total'] = dataa1[['jan', 'feb', 'mar', 'apr']].sum(axis=1)

# Atur style seaborn agar lebih rapi
sns.set_theme(style="whitegrid")

# Bar Chart
plt.figure(figsize=(8, 5)) # Ukuran kompak (lebar 8, tinggi 5 inci)
# Urutkan data dari total tertinggi
df_sorted = dataa1.sort_values('total', ascending=False)
sns.barplot(data=df_sorted, x='nama', y='total', palette='viridis')
plt.title('Total Penjualan per Orang', fontsize=12)
plt.xlabel('Nama Karyawan')
plt.ylabel('Total Penjualan')
plt.xticks(rotation=45) # Miringkan label agar tidak tumpang tindih
plt.tight_layout()
plt.show()
```

```python
# Pie chart
plt.figure(figsize=(5, 5)) # Ukuran persegi agar pie chart tidak gepeng

# Hitung total penjualan per bulan
total_per_bulan = dataa1[['jan', 'feb', 'mar', 'apr']].sum()
plt.pie(total_per_bulan, labels=total_per_bulan.index, autopct='%1.1f%%',
        startangle=140, colors=sns.color_palette('muted'))
plt.title('Proporsi Penjualan per Bulan', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# stack plot
plt.figure(figsize=(8,5))

dataa1.set_index('nama')[['jan', 'feb', 'mar', 'apr']].plot(
    kind='bar',
    stacked=True,
    colormap='Pastel1',
    ax=plt.gca()
)

plt.title('Komposisi Penjualan Bulanan per Orang (Stacked)', fontsize=12)
plt.xlabel('Nama Karyawan')
plt.ylabel('Jumlah Penjualan')
plt.xticks(rotation=45)
plt.legend(title='Bulan')
plt.tight_layout()
plt.show()
```

```python
# Histogran
plt.figure(figsize=(6, 4))

sns.histplot(dataa1['total'], bins=5, kde=True, color='teal')

plt.title('Distribusi Frekuensi Total Penjualan', fontsize=12)
plt.xlabel('Range Total Penjualan')
plt.ylabel('Jumlah Karyawan (Frekuensi)')
plt.tight_layout()
plt.show()
```

```python
# scatter
plt.figure(figsize=(6,4))

sns.scatterplot(data=dataa1, x='jan', y='feb', s=100, color='white', hue='nama')

for i in range(dataa1.shape[0]):
  plt.text(dataa1['jan'].iloc[i] + 5, dataa1['feb'].iloc[i], dataa1['nama'].iloc[i])

plt.title('Korelasi Penjualan: Januari vs Februari', fontsize=12)
plt.xlabel('Penjualan Januari')
plt.ylabel('Penjualan Februari')
plt.tight_layout()
plt.show()
```

```python
# Time Series Plot
plt.figure(figsize=(7, 4))

rata_bulanan = dataa1[['jan', 'feb', 'mar', 'apr']].mean()

sns.lineplot(x=rata_bulanan.index, y=rata_bulanan.values, marker='o')

plt.title('Rata-rata penjualan tim (Time Series)', fontsize=12)
plt.xlabel('Bulan')
plt.ylabel('Rata-rata Penjualan')
plt.tight_layout()
plt.show()
```
